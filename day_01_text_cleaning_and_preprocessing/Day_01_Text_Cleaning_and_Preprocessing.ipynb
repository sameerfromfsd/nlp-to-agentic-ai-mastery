## Topic 1: Text Cleaning & Normalization
**Objective:** Transform raw, noisy real-world text into sanitized linguistic data using regex, HTML parsers, and normalization rules.

```python
import re
import html
import unicodedata
print("Environment ready for Text Cleaning!")
```

```python
raw_samples = [
    "<p>Check out our brand-new AI agent at https://openai.com/chatgpt !! Contact help@ai.org for support. 🚀🔥</p>",
    "<div>The battery life on this laptop is absolute trash... &amp; customer service was terrible! 😡👎</div>",
    "Café au lait was ordered by naïve developers at résumé screening. Cost: $14.50."
]

for idx, sample in enumerate(raw_samples, 1):
    print(f"Sample {idx}: {sample}\n")
```

```python
def sanitize_text(text: str, handle_emojis: str = "keep") -> str:
    """
    Sanitizes raw text:
    1. Decodes HTML entities (&amp; -> &)
    2. Removes HTML tags (<p> -> '')
    3. Removes URLs
    4. Removes Emails
    5. Normalizes Unicode accents (NFKD)
    6. Manages emojis ('strip' or 'keep')
    """
    # 1. Decode HTML entities
    text = html.unescape(text)
    
    # 2. Strip HTML tags
    text = re.sub(r'<[^>]+>', '', text)
    
    # 3. Strip URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    
    # 4. Strip Emails
    text = re.sub(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', '', text)
    
    # 5. Unicode normalization (converts characters like 'é' to canonical 'e' + combining accent)
    text = unicodedata.normalize('NFKD', text)
    
    # 6. Emoji Handling
    if handle_emojis == "strip":
        # Remove characters outside standard ASCII/Latin ranges
        text = text.encode('ascii', 'ignore').decode('utf-8')
    
    # 7. Normalize consecutive spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

print("--- TESTING SANITIZATION (KEEP EMOJIS) ---")
for s in raw_samples:
    print(sanitize_text(s, handle_emojis="keep"))

print("\n--- TESTING SANITIZATION (STRIP EMOJIS) ---")
for s in raw_samples:
    print(sanitize_text(s, handle_emojis="strip"))
```

## Topic 1.3: Contraction Expansion
**Purpose:** Convert contractions (e.g., "don't", "can't") into their full grammatical form ("do not", "cannot") so models can clearly detect negations and verbs.

```python
# A simple lookup dictionary of common English contractions
CONTRACTION_MAP = {
    "don't": "do not",
    "can't": "cannot",
    "won't": "will not",
    "i'm": "i am",
    "it's": "it is",
    "they're": "they are",
    "we'll": "we will",
    "you've": "you have",
    "didn't": "did not",
    "wasn't": "was not"
}

def expand_contractions(text: str) -> str:
    """
    Replaces contractions with their expanded versions.
    """
    words = text.split()
    expanded_words = []
    
    for word in words:
        # Check if the lowercased word exists in our dictionary
        word_lower = word.lower()
        if word_lower in CONTRACTION_MAP:
            expanded_words.append(CONTRACTION_MAP[word_lower])
        else:
            expanded_words.append(word)
            
    return " ".join(expanded_words)

# Test it on a sample sentence
test_sentence = "I can't believe it's working and we'll learn NLP together!"
result = expand_contractions(test_sentence)

print("BEFORE:", test_sentence)
print("AFTER: ", result)
```
